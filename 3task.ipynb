```python
from collections import Counter, defaultdict
```

### Integer Square Root

```python
def my_sqrt(x: int) -> int:
    l, r = 0, x
    res = 0
    while l <= r:
        m = (l + r) // 2
        if m * m <= x:
            res = m
            l = m + 1
        else:
            r = m - 1
    return res

my_sqrt(8), my_sqrt(16), my_sqrt(0), my_sqrt(2147395599)
```

### Two Copiers

```python
def min_copy_time(n: int, x: int, y: int) -> int:
    if n == 1:
        return min(x, y)
    l, r = 0, (n - 1) * max(x, y)
    while l < r:
        m = (l + r) // 2
        if m // x + m // y >= n - 1:
            r = m
        else:
            l = m + 1
    return l + min(x, y)

min_copy_time(4, 1, 1), min_copy_time(5, 1, 2)
```

### Feed Animals

```python
def feed_animals(animals: list[int], food: list[int]) -> int:
    animals, food = sorted(animals), sorted(food)
    i = 0
    for f in food:
        if i < len(animals) and f >= animals[i]:
            i += 1
    return i

feed_animals([3, 4, 7], [8, 1, 2]), feed_animals([1, 2, 2], [7, 1])
```

### Find the Difference

```python
def find_the_difference_count(a: str, b: str) -> str:
    cnt = Counter(a)
    for ch in b:
        if cnt[ch] == 0:
            return ch
        cnt[ch] -= 1
    return ''

def find_the_difference(a: str, b: str) -> str:
    res = 0
    for ch in a + b:
        res ^= ord(ch)
    return chr(res)

find_the_difference('abcd', 'abcde'), find_the_difference('aabc', 'abaca'), find_the_difference_count('aabc', 'abaca')
```

### Two Sum

```python
def two_sum(nums: list[int], target: int) -> list[int]:
    seen = set()
    for num in nums:
        if target - num in seen:
            return [target - num, num]
        seen.add(num)
    return []

two_sum([3, 9, 11, 8, 16, 19, 21, 18], 25), two_sum([1, 2, 4], 10)
```

### Group Anagrams

```python
def group_anagrams(strs: list[str]) -> list[list[str]]:
    groups = defaultdict(list)
    for s in strs:
        key = [0] * 26
        for ch in s:
            key[ord(ch) - ord('a')] += 1
        groups[tuple(key)].append(s)
    return list(groups.values())

group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat'])
```
