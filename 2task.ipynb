```python
from collections import deque

class ListNode:
    def __init__(self, val: int = 0, next: 'ListNode | None' = None):
        self.val = val
        self.next = next
```

### Linked List Cycle

```python
def has_cycle(head: ListNode | None) -> bool:
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False
```

### Reverse Linked List

```python
def reverse_list(head: ListNode | None) -> ListNode | None:
    prev, cur = None, head
    while cur:
        cur.next, prev, cur = prev, cur, cur.next
    return prev
```

### Middle of the Linked List

```python
def middle_node(head: ListNode | None) -> ListNode | None:
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow
```

### Remove Linked List Elements

```python
def remove_elements(head: ListNode | None, val: int) -> ListNode | None:
    dummy = ListNode(0, head)
    prev, cur = dummy, head
    while cur:
        if cur.val == val:
            prev.next = cur.next
        else:
            prev = cur
        cur = cur.next
    return dummy.next
```

### Remove Nth Node From End of List

```python
def remove_nth_from_end(head: ListNode | None, n: int) -> ListNode | None:
    dummy = ListNode(0, head)
    slow = fast = dummy
    for _ in range(n + 1):
        fast = fast.next
    while fast:
        slow = slow.next
        fast = fast.next
    slow.next = slow.next.next
    return dummy.next
```

### Is Subsequence

```python
def is_subsequence_queue(a: str, b: str) -> bool:
    q = deque(a)
    for ch in b:
        if q and q[0] == ch:
            q.popleft()
    return not q

def is_subsequence(a: str, b: str) -> bool:
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            i += 1
        j += 1
    return i == len(a)

is_subsequence('abd', 'uabqd'), is_subsequence('abdr', 'uabqd')
```

### Valid Palindrome

```python
def is_palindrome_stack(s: str) -> bool:
    stack = list(s)
    for ch in s:
        if ch != stack.pop():
            return False
    return True

def is_palindrome(s: str) -> bool:
    l, r = 0, len(s) - 1
    while l < r:
        if s[l] != s[r]:
            return False
        l += 1
        r -= 1
    return True

is_palindrome('abracadabra'), is_palindrome('abracarba')
```

### Sort Stack Using Auxiliary Stack

```python
def sort_stack(nums: list[int]) -> list[int]:
    stack = []
    while nums:
        cur = nums.pop()
        while stack and stack[-1] > cur:
            nums.append(stack.pop())
        stack.append(cur)
    return stack

sort_stack([8, 9, 0, 6, 6, 1, 11, 4])
```

### Merge Two Sorted Lists

```python
def merge_two_lists(l1: ListNode | None, l2: ListNode | None) -> ListNode | None:
    dummy = cur = ListNode()
    while l1 and l2:
        if l1.val < l2.val:
            cur.next, l1 = l1, l1.next
        else:
            cur.next, l2 = l2, l2.next
        cur = cur.next
    cur.next = l1 or l2
    return dummy.next
```
