### Two Sum

```python
def two_sum(nums: list[int], target: int) -> list[int]:
    l, r = 0, len(nums) - 1
    while l < r:
        s = nums[l] + nums[r]
        if s == target:
            return [l, r]
        if s < target:
            l += 1
        else:
            r -= 1
    return []

two_sum([3, 8, 9, 11, 16, 18, 19, 21], 25)
```

### Reverse Array

```python
def reverse_array(nums: list[int]) -> list[int]:
    l, r = 0, len(nums) - 1
    while l < r:
        nums[l], nums[r] = nums[r], nums[l]
        l += 1
        r -= 1
    return nums

reverse_array([1, 2, 3, 4])
```

### Rotate Array

```python
def rotate_array(nums: list[int], k: int) -> list[int]:
    def rev(l: int, r: int) -> None:
        while l < r:
            nums[l], nums[r] = nums[r], nums[l]
            l += 1
            r -= 1

    n = len(nums)
    k %= n
    rev(0, n - 1)
    rev(0, k - 1)
    rev(k, n - 1)
    return nums

rotate_array([1, 2, 3, 4, 5, 6, 7], 8)
```

### Merge Two Sorted Arrays

```python
def merge_sorted(a: list[int], b: list[int]) -> list[int]:
    res = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1
    res.extend(a[i:])
    res.extend(b[j:])
    return res

merge_sorted([2, 4, 6, 8], [1, 3, 4, 4, 11])
```

### Merge Sorted Arrays In-Place

```python
def merge_in_place(a: list[int], b: list[int]) -> list[int]:
    p1 = len(a) - len(b) - 1
    p2 = len(b) - 1
    p3 = len(a) - 1

    while p2 >= 0:
        if p1 >= 0 and a[p1] > b[p2]:
            a[p3] = a[p1]
            p1 -= 1
        else:
            a[p3] = b[p2]
            p2 -= 1
        p3 -= 1
    return a

merge_in_place([3, 8, 10, 11, 0, 0, 0], [1, 7, 9])
```

### Minimum Size Subarray Sum

```python
def min_subarray_len(nums: list[int], target: int) -> int:
    min_len = float('inf')
    l = cur_sum = 0

    for r, val in enumerate(nums):
        cur_sum += val
        while cur_sum >= target:
            min_len = min(min_len, r - l + 1)
            cur_sum -= nums[l]
            l += 1

    return 0 if min_len == float('inf') else int(min_len)

min_subarray_len([2, 3, 1, 2, 4, 3], 9)
```

### Sort Binary Array

```python
def sort_binary(nums: list[int]) -> list[int]:
    l, r = 0, len(nums) - 1
    while l < r:
        if nums[l] == 1:
            nums[l], nums[r] = nums[r], nums[l]
            r -= 1
        else:
            l += 1
    return nums

sort_binary([0, 1, 1, 0, 0, 1, 0, 0, 1, 1])
```

### Sort Colors

```python
def sort_colors(nums: list[int]) -> list[int]:
    low, mid, high = 0, 0, len(nums) - 1
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
    return nums

sort_colors([1, 2, 0, 0, 1, 1, 1, 0, 2, 2, 2, 1])
```

### Move Even Numbers First

```python
def even_first(nums: list[int]) -> list[int]:
    pos = 0
    for i in range(len(nums)):
        if nums[i] % 2 == 0:
            nums[i], nums[pos] = nums[pos], nums[i]
            pos += 1
    return nums

even_first([1, 1, 1, 2, 3, 3, 4, 4, 4, 8, 7, 5, 0])
```

### Move Zeroes

```python
def move_zeroes(nums: list[int]) -> list[int]:
    pos = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[i], nums[pos] = nums[pos], nums[i]
            pos += 1
    return nums

move_zeroes([0, 0, 1, 0, 3, 12]), move_zeroes([0, 0, 0, 18, 16, 0, 0, 77, 99])
```
