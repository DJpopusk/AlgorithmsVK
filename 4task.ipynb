```python
from collections import deque

class TreeNode:
    def __init__(self, val: int = 0, left: 'TreeNode | None' = None, right: 'TreeNode | None' = None):
        self.val = val
        self.left = left
        self.right = right
```

### Build Binary Tree From Array

```python
def build_tree(arr: list[int | None], i: int = 0) -> TreeNode | None:
    if i >= len(arr) or arr[i] is None:
        return None
    root = TreeNode(arr[i])
    root.left = build_tree(arr, 2 * i + 1)
    root.right = build_tree(arr, 2 * i + 2)
    return root

def to_list(root: TreeNode | None) -> list[int | None]:
    res = {}
    q = deque([(root, 0)])
    while q:
        node, i = q.popleft()
        if node:
            res[i] = node.val
            q.append((node.left, 2 * i + 1))
            q.append((node.right, 2 * i + 2))
    return [res.get(i) for i in range(max(res) + 1)] if res else []

to_list(build_tree([8, 9, 11, 7, 16, 3, 1])), to_list(build_tree([3, 7, 8, None, 9, 6, 9]))
```

### Symmetric Tree

```python
def is_symmetric_bfs(root: TreeNode | None) -> bool:
    level = [root]
    while level:
        n = len(level)
        for i in range(n // 2):
            a, b = level[i], level[n - i - 1]
            if a is None and b is None:
                continue
            if a is None or b is None or a.val != b.val:
                return False
        level = [child for node in level if node for child in (node.left, node.right)]
    return True

def is_symmetric(root: TreeNode | None) -> bool:
    def mirror(a: TreeNode | None, b: TreeNode | None) -> bool:
        if a is None and b is None:
            return True
        if a is None or b is None or a.val != b.val:
            return False
        return mirror(a.left, b.right) and mirror(a.right, b.left)

    return mirror(root.left, root.right) if root else True

t1 = build_tree([3, 8, 8, 9, 6, 6, 9])
t2 = build_tree([3, 7, 8, None, 9, 6, 9])
is_symmetric(t1), is_symmetric(t2), is_symmetric_bfs(t1), is_symmetric_bfs(t2)
```

### Minimum Depth of Binary Tree

```python
def min_depth_dfs(root: TreeNode | None) -> int:
    if root is None:
        return 0
    if root.left is None:
        return 1 + min_depth_dfs(root.right)
    if root.right is None:
        return 1 + min_depth_dfs(root.left)
    return 1 + min(min_depth_dfs(root.left), min_depth_dfs(root.right))

def min_depth(root: TreeNode | None) -> int:
    if root is None:
        return 0
    q = deque([(root, 1)])
    while q:
        node, depth = q.popleft()
        if node.left is None and node.right is None:
            return depth
        if node.left:
            q.append((node.left, depth + 1))
        if node.right:
            q.append((node.right, depth + 1))
    return 0

t = build_tree([11, 8, 16, 2, 9, 6, None, 7, None, None, 9])
min_depth(t), min_depth_dfs(t), min_depth(build_tree([1, 2, None, 3]))
```

### Product of Min and Max in BST

```python
def min_max_product_tree(root: TreeNode | None) -> int:
    if root is None:
        return -1
    lo = hi = root
    while lo.left:
        lo = lo.left
    while hi.right:
        hi = hi.right
    return lo.val * hi.val

def min_max_product(data: list[int | None]) -> int:
    if not data or data[0] is None:
        return -1
    lo = 0
    while 2 * lo + 1 < len(data) and data[2 * lo + 1] is not None:
        lo = 2 * lo + 1
    hi = 0
    while 2 * hi + 2 < len(data) and data[2 * hi + 2] is not None:
        hi = 2 * hi + 2
    return data[lo] * data[hi]

bst = [16, 12, 18, 11, 15, 17, 21, None, None, None, None, None, None, 19, 24]
min_max_product(bst), min_max_product_tree(build_tree(bst)), min_max_product([8, 3, 10, 1, 6, None, 14])
```

### Same Tree

```python
def is_same_tree(a: TreeNode | None, b: TreeNode | None) -> bool:
    if a is None and b is None:
        return True
    if a is None or b is None or a.val != b.val:
        return False
    return is_same_tree(a.left, b.left) and is_same_tree(a.right, b.right)

a = build_tree([16, 12, 18, None, 15, 17])
b = build_tree([16, 12, 18, None, 15, 17])
c = build_tree([16, 11, 18, None, 15, None, 21])
is_same_tree(a, b), is_same_tree(a, c)
```
